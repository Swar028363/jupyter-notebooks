# Lab 5: Data Visualization and Interpretation using Seaborn

## Imports

```python
import numpy as np 
import pandas as pd 
import seaborn as sns 
import matplotlib.pyplot as plt
```

## Dataset 1: Penguins Dataset

```python
df_p = pd.read_csv("datasets/penguins_lter.csv")
df_p.head()
```

### 1. Catplot

1. Create a bar catplot to compare average body mass across species and sex.
2. Title: "Average Body Mass by Species and Sex"
3. X-axis: Species
4. Y-axis: Body Mass (g)
5. Hue: Sex

```python
g = sns.catplot(data=df_p, x="Species", y="Body Mass (g)", hue="Sex", kind="bar")
g.set_axis_labels("Species", "Body Mass (g)")
g.fig.suptitle("Average Body Mass by Species and Sex")
g.fig.autofmt_xdate()
g.fig.savefig("plots/lab5/catplot.png")
plt.show()
```

### 2. Stripplot

1. Visualize flipper length distribution by species and sex.
2. Title: "Flipper Length Distribution by Species and Sex"
3. X-axis: Species
4. Y-axis: Flipper Length (mm)
5. Hue: Sex

```python
sns.stripplot(
    data=df_p,
    x="Species",
    y="Flipper Length (mm)",
    hue="Sex"
)

plt.title("Flipper Length Distribution by Species and Sex")
plt.xlabel("Species")
plt.ylabel("Flipper Length (mm)")
plt.xticks(rotation=45)
plt.savefig("plots/lab5/stripplot.png")
plt.show()
```

### 3. Heatmap

1. Display correlation matrix of numerical features (bill length, bill depth, flipper length, body, mass)
2. Annotate values, use colorbar to represent correlation strength.
3. Title: "Correlation Heatmap of Penguin Features"

```python
corr = df_p[["Culmen Length (mm)", "Culmen Depth (mm)", "Flipper Length (mm)", "Body Mass (g)"]].corr()
sns.heatmap(corr, annot=True, cmap="coolwarm", cbar=True)
plt.title("Correlation Heatmap of Penguin Features")
plt.savefig("plots/lab5/heatmap.png")
plt.show()
```

### 4. Clustermap

1. Create a clustered heatmap to group features based on similarity.
2. Title: "Clustered Correlation Map of Penguin Features"

```python
corr = df_p[["Culmen Length (mm)", "Culmen Depth (mm)", "Flipper Length (mm)", "Body Mass (g)"]].corr()
g = sns.clustermap(corr, cmap="coolwarm", annot=True)
g.fig.suptitle("Clustered Correlation Map of Penguin Features", y=1.02)
plt.savefig("plots/lab5/clustermap.png")
plt.show()
```

## Dataset 2: Tips Dataset

```python
df_t = pd.read_csv("datasets/tip.csv")
df_t.head()
```

### 5. Box Plot

1. Compare total bill distribution across days.
2. Title: "Total Bill Distribution by Day"
3. X-axis: Day
4. Y-axis: Total Bill

```python
sns.boxplot(
    data=df_t,
    x="day",
    y="total_bill"
)

plt.title("Total Bill Distribution by Day")
plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.savefig("plots/lab5/boxplot.png")
plt.show()
```

### 6. Violin Plot

1. Show tip distribution by smoker status.
2. Title: "Tip Distribution by Smoker Status"
3. X-axis: Smoker
4. Y-axis: Tip

```python
sns.violinplot(
    data=df_t,
    x="smoker",
    y="tip"
)

plt.title("Tip Distribution by Smoker Status")
plt.xlabel("Smoker")
plt.ylabel("Tip")
plt.savefig("plots/lab5/violinplot.png")
plt.show()
```

### 7. KDE Plot

1. Plot density of total bill amounts split by gender.
2. Title: "Density of Total Bill by Gender"
3. X-axis: Total Bill
4. Hue: Gender

```python
sns.kdeplot(
    data=df_t,
    x="total_bill",
    hue="sex",
    fill=True
)

plt.title("Density of Total Bill by Gender")
plt.xlabel("Total Bill")
plt.savefig("plots/lab5/kdeplot.png")
plt.show()
```

### 8. Joint Plot

1. Explore relationship between total bill and tip.
2. Title: "Total Bill vs Tip Relationship"
3. X-axis: Total Bill
4. Y-axis: Tip
5. Hue: Gender

```python
g = sns.jointplot(
    data=df_t,
    x="total_bill",
    y="tip",
    hue="sex"
)

g.fig.suptitle("Total Bill vs Tip Relationship",y=1.02)
plt.savefig("plots/lab5/jointplot.png")
plt.show()
```

## Dataset 3: IRIS Dataset

```python
df_i = pd.read_csv("datasets/Iris.csv")
df_i.head()
```

### 9. Pairplot

1. Create pairwise relationships among all numeric features.
2. Title: "Pairwise Relationships in Iris Dataset"
3. Hue: Species

```python
g = sns.pairplot(
    df_i,
    hue="Species"
)

g.fig.suptitle("Pairwise Relationships in Iris Dataset", y=1.02)
plt.savefig("plots/lab5/pairplot.png")
plt.show()
```

### 10. Swarm Plot

1. Visualize petal length distribution across species.
2. Title: "Petal Length Distribution by Species"
3. X-axis: Species
4. Y-axis: Petal Length

```python
sns.swarmplot(
    data=df_i,
    x="Species",
    y="PetalLengthCm"
)

plt.title("Petal Length Distribution by Species")
plt.xlabel("Species")
plt.ylabel("Petal Length")
plt.savefig("plots/lab5/swarmplot.png")
plt.show()
```

### 11. Bubble Chart (Scatterplot with size)

1. Sepal length vs sepal width
2. bubble size = petal length
3. hue = species.
4. Title: "Sepal vs Petal Dimensions in Iris"

```python
sns.scatterplot(
    data=df_i,
    x="SepalLengthCm",
    y="SepalWidthCm",
    size="PetalLengthCm",
    hue="Species",
    sizes=(20, 200)
)

plt.title("Sepal vs Petal Dimensions in Iris")
plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.legend(bbox_to_anchor=(1.05, 1), loc="upper left")
plt.savefig("plots/lab5/bubble_chart.png")
plt.show()
```
