```python
fig, ax = plt.subplots(2, 2, figsize=(10, 10))

# ---------------- Plot 1: Square ----------------
x_sq = [2, 2, 5, 5, 2]
y_sq = [2, 5, 5, 2, 2]
ax[0, 0].plot(x_sq, y_sq, color='green', linewidth=2)
ax[0, 0].grid(True)
ax[0, 0].set_xlim(0, 10)
ax[0, 0].set_ylim(0, 10)
ax[0, 0].set_title("Plot 1: Square")

# ---------------- Plot 2: Circle and Cross ----------------
theta = np.linspace(0, 2*np.pi, 100)
# Circle centered at (5, 5)
x_c = 5 + 2*np.cos(theta)
y_c = 5 + 2*np.sin(theta)
ax[0, 1].plot(x_c, y_c)
# Cross centered
ax[0, 1].plot([5, 5], [3, 7]) # vertical line
ax[0, 1].plot([3, 7], [5, 5]) # horizontal line
ax[0, 1].grid(False)
ax[0, 1].set_xlim(0, 10)
ax[0, 1].set_ylim(0, 10)
ax[0, 1].set_aspect('equal')
ax[0, 1].set_title("Plot 2: Circle & Cross")

# ---------------- Plot 3: Spiral ----------------
theta = np.linspace(0, 4*np.pi, 200)
r = theta
x_s = r * np.cos(theta)
y_s = r * np.sin(theta)
ax[1, 0].plot(x_s, y_s)
ax[1, 0].grid(False)
ax[1, 0].set_title("Plot 3: Spiral")

# ---------------- Plot 4: Triangle ----------------
x_t = [2, 6, 10, 2]
y_t = [2, 10, 2, 2]
ax[1, 1].plot(x_t, y_t, color='red')
ax[1, 1].grid(False)
ax[1, 1].set_xlim(0, 12)
ax[1, 1].set_ylim(0, 12)
ax[1, 1].set_title("Plot 4: Triangle")
plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()
```

## Task 3: Shapes using patches (ticks and tick labels(major and minor))

```python
fig, ax = plt.subplots()
x_sq = [2, 2, 5, 5, 2]
y_sq = [2, 5, 5, 2, 2]

# Plot square
ax.plot(x_sq, y_sq, color='yellow')
ax.set_xlim(0, 10)
ax.set_ylim(0, 10)

# Major ticks
ax.set_xticks(range(0, 11, 2))
ax.set_yticks(range(0, 11, 2))

# Enable minor ticks
ax.minorticks_on()
ax.grid(True, which='major', linestyle='--', linewidth=1, color='blue')
ax.grid(True, which='minor', linestyle=':', linewidth=1.5, color='gray')
ax.tick_params(top=True, bottom=True, left=True, right=True)
ax.tick_params(axis='both', which='major',direction='inout', length=10, width=1.5)
ax.set_title("Major and Minor Grid", fontname="Times New Roman")
plt.show()
```

## Task 4: Inset plotting (plot + legend + title)

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.axes_grid1.inset_locator import inset_axes, mark_inset

# Generate data
x = np.linspace(0, 100, 6000)
y = 0.05 * x

# Add glitch between 47 and 48
for i in range(len(x)):
    if 47 < x[i] < 48:
        y[i] += np.sin(10 * x[i])

# Main figure
fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(x, y, label="Line with glitch")

# Mark zoom region
x1, x2 = 46.8, 48.2
ax.axvline(x1, color="red")
ax.axvline(x2, color="red")
ax.axvspan(x1, x2, alpha=0.3)

# Create inset
axins = inset_axes(
    ax,
    width="35%",
    height="45%",
    loc="upper right",
    borderpad=1.2
)

# Extract zoomed data
mask = (x > x1) & (x < x2)
x_zoom = x[mask]
y_zoom = y[mask]

axins.plot(x_zoom, y_zoom, label="Glitch")

# Connect inset to main plot
mark_inset(ax, axins, loc1=2, loc2=4,
           fc="none", ec="black", linewidth=1)

# Labels & formatting
ax.set_xlabel("Time")
ax.set_ylabel("Signal Amplitude")
ax.set_title("Zoomed Plot", loc="left", pad=10)

ax.grid(True, linestyle="--", alpha=0.25)
ax.set_xlim(0, 100)
ax.set_ylim(0, 5)

# Styled legend
ax.legend(
    bbox_to_anchor=(1.1, 1.1),
    fontsize=10,
    title="Lines",
    frameon=True,
    facecolor="red",
    edgecolor="black",
    handlelength=5.0,
    handleheight=2.0,
    markerscale=0.2
)

plt.show()
```

## Task 6: Pie + Donut

### Pie Chart

```python
data = [10, 20, 20, 25, 25]
labels = ["M1", "M2", "M3", "M4", "M5"]
colors = ["yellow", "orange", "blue", "green", "pink"]

plt.figure(figsize=(6,6))

plt.pie(
    data,
    labels=labels,
    colors=colors,
    startangle=90,
    counterclock=False,
    autopct="%1.1f%%"
)

plt.title("Pie chart", fontsize=14)
plt.axis("equal") 
plt.show()
```

### Donut Chart

```python
data = [10, 20, 20, 25, 25]
labels = ["M1", "M2", "M3", "M4", "M5"]
colors = ["yellow", "orange", "blue", "green", "pink"]

fig, ax = plt.subplots(figsize=(6,6))

# Light grey background
fig.patch.set_facecolor("#f2f2f2")
ax.set_facecolor("#f2f2f2")

# Donut chart
ax.pie(
    data,
    labels=labels,
    colors=colors,
    startangle=90,
    counterclock=False,
    autopct="%1.1f%%",
    pctdistance=0.8, 
    wedgeprops=dict(width=0.45) 
)

ax.set_title("Donut chart", fontsize=14)

# Center text
ax.text(
    0, 0,
    "Module Weights",
    ha='center',
    va='center',
    fontsize=10,
    color="black"
)

ax.axis("equal")

plt.show()
```

## Task 7: 3D plotting

### 3D Bar Chart

```python
from mpl_toolkits.mplot3d import Axes3D

x = [1,2,3,4]
z = [160, 80, 75, 65]
y = 2

x = np.array(x)
z_base = 2

dx = 0.5
dy = 0.5
dz = z

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

ax.bar3d(x,y, z_base, dx,dy,dz)
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")
ax.set_xlim(0,10)
ax.set_ylim(0,10)
ax.set_title("3d Bar Plot")
ax.view_init(elev=25, azim=45)
plt.show()
```

### Ploting Equations

```python
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)

X, Y = np.meshgrid(x, y)
Z = X**2 + Y**2

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")
ax.set_title("Ploting Equations")
ax.plot_surface(X,Y,Z, cmap = 'viridis')
ax.view_init(elev=25, azim=45)
plt.show()
```

## Task 8: Animation

```python
from matplotlib.animation import FuncAnimation

tz = ZoneInfo("Asia/Kolkata")

def angle_rad(angle, length):
    theta = np.radians(90 - angle)
    return length * np.cos(theta), length * np.sin(theta)

fig, ax = plt.subplots(figsize=(6, 6))

circle = plt.Circle((0, 0), 1, fill=False, linewidth=2)
ax.add_artist(circle)

ax.set_xlim(-1.1, 1.1)
ax.set_ylim(-1.1, 1.1)
ax.set_aspect('equal')
ax.axis('off')

hour_hand, = ax.plot([], [], linewidth=6, color='black')
minute_hand, = ax.plot([], [], linewidth=3, color='blue')
second_hand, = ax.plot([], [], linewidth=1, color='red')

title = ax.set_title("")

def update(frame):
    now = datetime.now(tz)

    hour = now.hour % 12
    minute = now.minute
    second = now.second

    hour_angle = (hour + minute / 60) * 30
    minute_angle = (minute + second / 60) * 6
    second_angle = second * 6

    hx, hy = angle_rad(hour_angle, 0.5)
    mx, my = angle_rad(minute_angle, 0.7)
    sx, sy = angle_rad(second_angle, 0.9)

    hour_hand.set_data([0, hx], [0, hy])
    minute_hand.set_data([0, mx], [0, my])
    second_hand.set_data([0, sx], [0, sy])

    title.set_text(f"Analog Clock - IST ({now.strftime('%H:%M:%S')})")

ani = FuncAnimation(fig, update, frames=360, interval=1000)

from IPython.display import HTML
HTML(ani.to_jshtml())
```
