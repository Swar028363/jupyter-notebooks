```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
sns.get_dataset_names()
```

```python
df = sns.load_dataset("flights")
df.head()
```

```python
fig, axes = plt.subplots(1,2, figsize=(12,5))
sns.histplot(df["passengers"], kde=True, ax=axes[0], color="blue")
axes[0].set_title("Distribution of Passengers")

sns.scatterplot(data=df, x="year", y="passengers", hue="month", ax=axes[1], palette="tab20")
axes[1].set_title("Year Vs Passangers")

plt.show()
```

## Task 2: Flights Dataset: Plot Analysis

### 1. how did the total nu,ber of airline passengers change from 1949 to 1960

```python
sns.lineplot(data=df, x="year", y="passengers", hue="month")
plt.title("Year Vs Passengers")
plt.grid()
plt.show()

sns.barplot(data=df, x="year", y="passengers", hue="month")

plt.title("Year Vs Passengers")
plt.show()
```

### 2. how did passenger count change over years for a single month

```python
for mon in df["month"].unique():
    df_mon = df[df["month"] == mon]
    sns.lineplot(data=df_mon, x="year", y="passengers")
    plt.title(f"Passengers {mon.title()} over Years")
    plt.grid()
    plt.show()
```

### 3. in a chossen year which month had a higher passenger counts?

```python
for yer in df["year"].unique():
    df_year = df[df["year"] == yer]
    sns.barplot(data=df_year, x="month", y="passengers")
    plt.title(f"Distribution of Passengers for year {yer}")
    plt.show()
```
