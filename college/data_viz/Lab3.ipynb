## Task: 1 

Write code for line chart to generate 70 numbers with X- axis and Y-axis is mx+c, were m=2 andc=0.3, Draw the four different line charts using matplotlib library. (Take Y1=X, Y2=X**2, Y3=X**3 and Y4=sqrt(X)).

````python
x = np.linspace(0, 2, 70)
m, c = 2, 0.3

y = m * x + c
y1 = x
y2 = x**2
y3 = x**3
y4 = np.sqrt(x)

plt.figure()
plt.plot(x, y, label=f"Y = {m} * X + {c}")
plt.plot(x, y1, label="Y = X")
plt.plot(x, y2, label="Y = X**2")
plt.plot(x, y3, label="Y = X**3")
plt.plot(x, y4, label="Y = sqrt(X)")
plt.title("Line Charts for Different Functions")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid()
plt.legend()
plt.savefig("plots/lab3/task1_line_charts.png")
plt.show()
````

X values range from 1 to 70.  
Multiple functions are plotted on the same graph.  
Legends and grid improve readability.  

## Task: 2

Plot the function y = x**2 - 4x + 3 for x ranging from -2 to 6. Identify and annotate the minimumpoint
of the graph. Add appropriate labels, title, and grid.

````python
x = np.linspace(-2, 6, 100)
y = (x**2) - (4 * x) + 3

x_min = 2
y_min = (x_min**2) - (4*x_min) + 3

plt.figure()
plt.plot(x, y, label="y = x**2 - 4x + 3")
plt.scatter(x_min, y_min)
plt.annotate(
    f"Minimum Point ({x_min}, {y_min})",
    xy=(x_min, y_min),
    xytext=(1, 2),
    arrowprops=dict(arrowstyle="->")
)
plt.title("Graph of y = x**2 - 4x + 3")
plt.xlabel("x")
plt.ylabel("y")
plt.grid()
plt.legend(loc="upper center")
plt.savefig("plots/lab3/task2_quadratic.png")
plt.show()
````

The parabola opens upwards.  
Minimum occurs at x = 2, y = -1.  
Annotation highlights the minimum point.  

## Task: 3

Plot sin(x) and cos(x) from 0 to 2π on the same graph. Use fill_between() to shade the area betweenthetwo curves where sin(x) is greater than cos(x). Add labels and a title.

````python
x = np.linspace(0, 2*np.pi, 100)

y_sin = np.sin(x)
y_cos = np.cos(x)

plt.figure()
plt.plot(x, y_sin, label="sin(x)")
plt.plot(x, y_cos, label="cos(x)")
plt.fill_between(
    x, y_sin, y_cos, 
    where=(y_sin > y_cos) | (y_sin < y_cos),
    alpha=0.3
)
plt.title("Sin and Cos with Shaded Region")
plt.xlabel("x")
plt.ylabel("Value")
plt.legend()
plt.savefig("plots/lab3/task3_sin_cos.png")
plt.show()
````

Both functions are plotted from 0 to 2π.  
fill_between() highlights the required region.

## Task: 5

Plot three mathematical functions on the same graph:  
1. y = x² (Red, Dashed Line)
2. y = 2x + 1 (Blue, Solid Line)
3. y = sqrt(x) (Green, Dotted Line)
   
Add a legend, labels, and title.

````python
x = np.linspace(0, 5, 100)

y1 = x**2
y2 = 2*x + 1
y3 = np.sqrt(x)

plt.figure()
plt.plot(x, y1, 'r--', label="y = x**2")
plt.plot(x, y2, 'b-', label="y = 2x + 1")
plt.plot(x, y3, 'g:', label="y = sqrt(x)")
plt.title("Multiple Mathematical Functions")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)
plt.savefig("plots/lab3/task5_multiple_functions.png")
plt.show()
````

Different line styles distinguish functions.  
Legend clearly identifies each function.  

## Task: 6

Create a dual-axis plot where: 
1. The primary y-axis represents the monthly rainfall (in mm) in 2023 as a bar chart.
2. The secondary y-axis represents the average temperature (in °C) as a line graph on the sameplot.
```Python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
rainfall = [78, 62, 55, 41, 32, 140, 300, 280, 150, 90, 85, 80]
temperature = [12, 14, 18, 24, 30, 34, 32, 31, 28, 22, 16, 13]
```

Label both y-axes and use different colors for bars and the line.

````python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
rainfall = [78, 62, 55, 41, 32, 140, 300, 280, 150, 90, 85, 80]
temperature = [12, 14, 18, 24, 30, 34, 32, 31, 28, 22, 16, 13]

fig, ax1 = plt.subplots()
ax1.bar(months, rainfall, color='skyblue')
ax1.set_ylabel("Rainfall (mm)")

ax2 = ax1.twinx()
ax2.plot(months, temperature, color='red', marker='o')
ax2.set_ylabel("Temperature (°C)")

plt.title("Monthly Rainfall and Average Temperature in 2023")
plt.savefig("plots/lab3/task6_dual_axis.png")
plt.show()
````

Bar chart represents rainfall.  
Line graph represents temperature.  
Dual y-axes allow comparison.  

## Task: 7

Draw a house using matplotlib line charts:

````python
roof_x = [1, 3, 5, 1]
roof_y = [4, 6, 4, 4]

walls_x = [1, 1, 5, 5, 1]
walls_y = [1, 4, 4, 1, 1]

door_x = [2, 2, 3, 3, 2]
door_y = [1, 3, 3, 1, 1]

win1_x = [4, 4, 4.5, 4.5, 4]
win1_y = [3, 3.5, 3.5, 3, 3]

win2_x = [4, 4, 4.5, 4.5, 4]
win2_y = [1.5, 2, 2, 1.5, 1.5]

plt.figure(figsize=(8, 8))
plt.plot(roof_x, roof_y, color="darkred", label="Roof")
plt.plot(walls_x, walls_y, color="darkred", label="Walls")
plt.plot(door_x, door_y, color="Blue", label="Door")
plt.plot(win1_x, win1_y, color="green", label="Window 1")
plt.plot(win2_x, win2_y, color="Purple", label="Window 2")
plt.xlabel("X-axis")
plt.grid(True)
plt.legend()
plt.savefig("plots/lab3/task7_house.png")
plt.show()
````

Lines are used to form the walls, roof, windows and door.  
Equal axis scaling maintains proportions.  

````python
roof_x = [1, 3, 5, 1]
roof_y = [4, 6, 4, 4]
roof = Polygon([
    [1,4], [3,6], [5,4]
], fill=False, color="darkred")
walls_x = [1, 1, 5, 5, 1]
walls_y = [1, 4, 4, 1, 1]

door_x = [2, 2, 3, 3, 2]
door_y = [1, 3, 3, 1, 1]

win1_x = [4, 4, 4.5, 4.5, 4]
win1_y = [3, 3.5, 3.5, 3, 3]

win2_x = [4, 4, 4.5, 4.5, 4]
win2_y = [1.5, 2, 2, 1.5, 1.5]

fig, ax = plt.subplots()
ax.add_patch(roof)
plt.plot(walls_x, walls_y, color="darkred", label="Walls")
plt.plot(door_x, door_y, color="Blue", label="Door")
plt.plot(win1_x, win1_y, color="green", label="Window 1")
plt.plot(win2_x, win2_y, color="Purple", label="Window 2")
plt.xlabel("X-axis")
plt.grid(True)
plt.legend()
plt.savefig("plots/lab3/task7_house.png")
plt.show()
````
