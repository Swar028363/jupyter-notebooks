## Program 1: Tensor Creation and Basic Operations

### Aim
To understand tensor creation and perform basic arithmetic operations using
TensorFlow

### Algorithm

1. Import TensorFlow library

```python
import tensorflow as tf
```

2. Create scalar, vector, and matrix tensors

```python
# Scalar
s1 = tf.constant(5)
s2 = tf.constant(3)

# Vector
v1 = tf.constant([1, 2, 3])
v2 = tf.constant([4, 5, 6])

# Matrix
m1 = tf.constant([[1, 2], [3, 4]])
m2 = tf.constant([[5, 6], [7, 8]])
```

3. Perform addition, multiplication, and matrix multiplication

```python
scaler_add = s1 + s2
vec_add = v1 + v2
mat_mul = tf.matmul(m1, m2)
```

4. Display results

```python
print("Scalar Addition:", scaler_add)
print("Vector Addition:", vec_add)
print("Matrix Multiplication:\n", mat_mul)
```

## Program 2: Linear Regression Using TensorFlow

### Aim
To implement a simple linear regression model using TensorFlow.

Algorithm

1. Import required libraries

```python
import tensorflow as tf
import numpy as np
```

2. Create input and output data

```python
x = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([2, 4, 6, 8, 10], dtype=float)
```

3. Define a linear model

```python
model = tf.keras.Sequential([
tf.keras.layers.Dense(units=1, input_shape=[1])
])
```

4. Compile the model with loss and optimizer

```python
model.compile(optimizer='sgd', loss='mean_squared_error')
```

5. Train the model

```python
model.fit(x, y, epochs=10, verbose=0)
```

6. Test the model

```python
print("Prediction for x=6:", model.predict(np.array([6])))
```

## Program 3: Simple Neural Network for Classification

### Aim
To build and train a neural network for binary classification.

### Algorithm

1. Import TensorFlow and dataset

```python
import tensorflow as tf
from sklearn.datasets import make_classification
```

```python
X, y = make_classification(n_samples=500, n_features=2, n_informative=2, n_redundant=0, n_classes=2)
X[:5], y[:5]
```

2. Create a sequential neural network

```python
model = tf.keras.Sequential()
```

3. Add dense layers with activation functions

```python
model.add(tf.keras.layers.Dense(8, activation='relu', input_shape=(2,)))
model.add(tf.keras.layers.Dense(1, activation='sigmoid'))
```

4. Compile the model

```python
model.compile(optimizer='adam',
loss='binary_crossentropy',
metrics=['accuracy'])
```

5. Train and evaluate the model

```python
model.fit(X, y, epochs=10, verbose=0)
loss, acc = model.evaluate(X, y, verbose=0)
print("Loss:", loss)
print("Accuracy:", acc)
```

5. Implement multi-class classification using Softmax activation

```python
X, y = make_classification(
    n_samples=500,
    n_features=10,
    n_informative=4,
    n_redundant=0,
    n_classes=4,
    n_clusters_per_class=1,
)

model = tf.keras.Sequential([
tf.keras.layers.Dense(8, activation='relu', input_shape=(10,)),
tf.keras.layers.Dense(16, activation='relu'),
tf.keras.layers.Dense(32, activation='tanh'),
tf.keras.layers.Dense(16, activation='relu'),
tf.keras.layers.Dense(4, activation='softmax')
])

model.compile(optimizer='adam',
loss='sparse_categorical_crossentropy',
metrics=['accuracy'])

model.fit(X, y, epochs=100, verbose=0)

loss, acc = model.evaluate(X, y, verbose=0)
print("Loss:", loss)
print("Accuracy:", acc)
```

## Exercise Questions

1. What is a tensor?

A Tensor is a N-dimensional Matrix:  

1.  A Scalar is a 0-dimensional tensor  
2.  A Vector is a 1-dimensional tensor  
3.  A Matrix is a 2-dimensional tensor  
4.  A n-dimentional tensor

A Tensor is a generalization of Vectors and Matrices to higher dimensions.  

2. Difference between NumPy arrays and TensorFlow tensors

NumPy arrays and TensorFlow tensors are both multi-dimensional arrays used for numerical computation in Python, but tensors are designed specifically for the demands of deep learning, offering GPU acceleration, automatic differentiation, and a computational graph. NumPy arrays are designed for general-purpose scientific computing on the CPU.

| Feature | NumPy Arrays | TensorFlow Tensors |
|------|-------------|-------------------|
| **Primary Use Case** | General scientific computing and data analysis. Works well with libraries like SciPy, Matplotlib, and Pandas. | Deep learning, machine learning model training, and distributed computing. |
| **Hardware Acceleration** | Primarily CPU-based. Requires additional libraries for limited GPU support. | Designed to run efficiently on CPUs, GPUs, TPUs, and distributed clusters. |
| **Automatic Differentiation** | No built-in support. Requires manual implementation or external libraries. | Built-in support for automatic differentiation, essential for backpropagation in neural networks. |
| **Execution Model** | Eager execution: operations run immediately, step by step. | Supports both eager execution and graph execution. |
| **Mutability** | Mutable: values can be modified after creation. | Immutable: tensors cannot be modified; new tensors are created for updates. |
| **Integration & Ecosystem** | Core of the Python scientific stack with a simple and lightweight API. | Part of the TensorFlow ecosystem with NumPy interoperability. |


3. What is an optimizer?

In deep learning, optimizers are crucial as algorithms that dynamically fine-tune a model’s parameters throughout the training process, aiming to minimize a predefined loss function. These specialized algorithms facilitate the learning process of neural networks by iteratively refining the weights and biases based on the feedback received from the data. Well-known optimizers in deep learning encompass Stochastic Gradient Descent (SGD), Adam, and RMSprop, each equipped with distinct update rules, learning rates, and momentum strategies, all geared towards the overarching goal of discovering and converging upon optimal model parameters, thereby enhancing overall performance.

4. Explain loss functions

A loss function is a mathematical way to measure how good or bad a model’s predictions are compared to the actual results.  

Loss functions are important because they:
1. Guide Model Training
2. Measure Performance
3. Affect learning behavior

5. What is backpropagation?

Backpropagation, short for Backward Propagation of Errors, is a key algorithm used to train neural networks by minimizing the difference between predicted and actual outputs. It works by propagating errors backward through the network, using the chain rule of calculus to compute gradients and then iteratively updating the weights and biases. Combined with optimization techniques like gradient descent, backpropagation enables the model to reduce loss across epochs and effectively learn complex patterns from data.
